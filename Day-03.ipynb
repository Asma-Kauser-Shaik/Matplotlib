# Multiple subplots

```python
import matplotlib.pyplot as plt
```

```python
# plot mutliple plots together 
# Method-1: using plt.subplot():good for less number of plots
# Method-2 : using plt.subplots():good large number of subplots
```

# Method-1: using plt.subplot():good for less number of plots

```python
# plt.subplot(nrows,ncols,index)
mobile_company = ["Realme","Samsung","Nokia","Lava","Apple","IQoo"]
likes_count = [15,11,18,34,14,8]
plt.subplot(2,3,1)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,3)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,4)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,6)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,5)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,2)
plt.pie(likes_count,labels = mobile_company)
plt.tight_layout() # use this if subplots are overlapping
plt.show()
```

```python
mobile_company = ["Realme","Samsung","Nokia","Lava","Apple","IQoo"]
likes_count = [15,11,18,34,14,8]
food_items = ["Sambar","pongal","Dosa","upma"]
likes = [22,34,56,2]
plt.subplot(2,3,1)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,3)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,4)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,2)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,5)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,6)
plt.barh(food_items,likes,color = "Yellow")
plt.tight_layout() # use this if subplots are overlapping
plt.show()
```

```python
mobile_company = ["Realme","Samsung","Nokia","Lava","Apple","IQoo"]
likes_count = [15,11,18,34,14,8]
food_items = ["chicken Biryani","Mutton Biryani","prawan fry","chickenfry"]
count = [20,30,10,40]
plt.subplot(2,3,6)
plt.pie(count,labels = food_items,autopct ='%1.1f%%',wedgeprops = dict(width = 0.6))
plt.subplot(2,3,1)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,3)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,4)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,2)
plt.pie(likes_count,labels = mobile_company)
plt.subplot(2,3,5)
plt.pie(likes_count,labels = mobile_company)
plt.tight_layout() # use this if subplots are overlapping
plt.show()
```

# Method-2 using plt.subplots():good for large of subplots

```python
Dinner = ["Biryani","chicken","Mutton","Roti","Lassi"]
percentage = [30,20,10,20,20]
fig,axes = plt.subplots(2,2 ,figsize = (10,10)) # (nrows,ncols,full image size)
for ax in axes.ravel():
    ax.pie(percentage,labels = Dinner,autopct = '%1.1f%%')
```

```python
# if we want to pass title for piecharts.
Dinner = ["Biryani","chicken","Mutton","Roti","Lassi"]
percentage = [30,20,10,20,20]
_,axes = plt.subplots(2,3 ,figsize = (10,10)) # (nrows,ncols,full image size)(we use _ if we dont want any thing)
titles = ["Asma","kauser","venkat","mani","sreeja","abc"]
print(axes)
axes_id = axes.ravel()
print(axes_id)
for title, ax in zip(titles,axes.ravel()):
    ax.pie(percentage,labels = Dinner,autopct = '%1.1f%%')
    ax.set_title(title)
```

```python
Dinner = ["Biryani","chicken","Mutton","Roti","Lassi"]
percentage = [30,20,10,20,20]
food_items = ["Sambar","pongal","Dosa","upma"]
likes = [22,34,56,2]
_,axes = plt.subplots(2,3 ,figsize = (10,10)) # (nrows,ncols,full image size)(we use _ if we dont want any thing)
titles = ["Asma","kauser","venkat","mani","sreeja","abc"]
for title, ax in zip(titles,axes.ravel()):
    if title == "Asma":
        ax.bar(food_items,likes)
        ax.set_title(title)
    else:
         ax.pie(percentage,labels = Dinner,autopct = '%1.1f%%')
         ax.set_title(title)
```

# 3D charts in matplotlib

```python
import numpy as np 
z = np.linspace(0,10,100)
x = np.sin(z)
y = np.cos(z)
fig = plt.figure()
ax = fig.add_subplot(111,projection = '3d')
ax.plot(x,y,z)
plt.show()
```

```python
import numpy as np 
z = np.linspace(0,10,100)
x = np.sin(z)
y = np.cos(z)
fig = plt.figure()
ax = fig.add_subplot(111,projection = '3d')
ax.scatter(x,y,z)
plt.show()
```
