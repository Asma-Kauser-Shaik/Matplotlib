# Scatter plot 

```python
import matplotlib.pyplot as plt
```

```python
# Relationship between two variables
# Types:
# Postive relation(both are same if increase or decreasing)
# Negative relation(one increase another decrease) # by reverse()
# No relation
```

```python
age = [5,6,7,8,9,10,11,12,13]
classes = [1,2,3,4,5,6,7,8,9]
plt.scatter(age,classes,s = 50,color="Black")
plt.title("Age VS Classes")
plt.xlabel("age")
plt.ylabel("classes")
```

```python
age = [5,6,7,8,9,10,11,12,13]
classes =[9,8,7,6,5,4,3,2,1]#classes.reverse() we using reverse also 
plt.scatter(age,classes)
```

# Pie chart 

```python
# catergorical data 
# no of catergorical data is less than  pie chart is used if is more than use bar chart.
# to repersent the data in form of percentage/porportion
```

```python
food_items = ["chicken Biryani","Mutton Biryani","prawan fry","chickenfry"]
likes_count = [20,30,10,40]
plt.pie(likes_count,labels = food_items,autopct ='%1.1f%%')
plt.title("food Items vs likes count")
plt.show()
plt.savefig("ds20_piechart.jpg")
plt.savefig("ds20_piechart.png")
plt.savefig("ds20_piechart.pdf")
```

# Donut chart

```python
# Same as piechart
# donut chart is like pie chart only but space at the center
# this space can be used for some extra information
```

```python
food_items = ["chicken Biryani","Mutton Biryani","prawan fry","chickenfry"]
likes_count = [20,30,10,40]
plt.pie(likes_count,labels = food_items,autopct ='%1.1f%%',wedgeprops = dict(width = 0.6))
plt.text(0,0,"Sales",ha = "center")
plt.title("food Items vs likes count")
plt.show()
```
