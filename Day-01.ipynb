# Bar chart

```python
# USE: used to shows the discrete values of categorical data

food_items = ["idli", "Dosa", "Vada", "Sambhar", "Puri"]
like_count = [2, 7, 3, 10, 12]

plt.figure(figsize = (3,2))

plt.bar(food_items, like_count, color = ["red", "blue", "orange", "green","purple"])

for i, count in enumerate(like_count):
    plt.text(i, count+0.2, count, ha = "center") # plt,text(x_index, y_index, value, position)

plt.xlabel("food items")
plt.ylabel("number of likes")
```

# Histogram

```python
# Histogram works on continues data
# Histogram is popularly used to find the shape of the data

import numpy as np

np.random.seed(42)
data = np.random.randn(1000) # normal distributed data : range = -3 to +3 ==> 99.7%

plt.figure(figsize = (6,4))

plt.hist(data, edgecolor = 'black', bins = 5)
plt.title("Histogram chart")
plt.xlabel("X-data")
plt.ylabel("y-data")
plt.show()
```

```python
# Interview Question : what is the difference between Bar chart and Histogram?
# ANS : Bar chart works on the catogorical data while histogram works on continuous data
# Bins in histogram  don't have space between them.
```

# Line chart

```python
# USE: to show trends over the time
# time series data

import pandas as pd
dates = pd.date_range(start = "2026-01-01", periods = 12, freq = "ME")
dates
```

```python
icecream_sales = [60, 70, 100, 20, 40, 50, 90, 120, 130, 140, 170, 190]
len(icecream_sales)
```

```python
plt.figure(figsize = (4,3))
plt.plot(dates,icecream_sales, marker = "o")

for d, ice in zip(dates,icecream_sales):
    plt.text(d, ice+5, ice, ha = "center") # plt,text(x_index, y_index, value, position)

plt.xticks(dates, dates.strftime("%b"), rotation = 45)
plt.show()
```

# Scatter plot

```python
# USE: To find the relationship between two variables
# e.g finding the relationship between age and salary

# type of relationships:
# 1. positive relation(both variable values increasing or both var value decreasing
# 2. negative relation(one increase other decrease)
# 3. no relation

age = [5, 8, 10, 15, 20, 25, 30, 40, 50, 68, 99]
salary = [2, 5, 10, 20, 50, 100, 200, 500, 1000, 2000, 3999]
salary.reverse()
print(salary)
len(salary)

plt.scatter(age, salary, s = 100, color = "black")

plt.title("Age VS Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
```

# Pie chart

```python
# USE: used to represent less number of categorical data
# To represent the data in form of percentage/proportion
# in pie chart we shows percentage of various categories

mobile_company = ["Realme", "Samsung", "Nokia", "Lava", "Apple", "IQoo", "OnePlus"]
market_share = [15, 25, 2, 3, 30, 5, 20]

plt.figure(figsize = (5,4))

plt.pie(market_share, labels = mobile_company, autopct = '%1.1f%%')
plt.title("Mobile market share")

plt.savefig("ds20_piechart.jpg")
plt.savefig("ds20_piechart.png")
plt.savefig("ds20_piechart.pdf")
```

```python
import matplotlib.pyplot as plt
```

```python
# plot multiple plots together

# Method-1 : using plt.subplot() : good for less number of plots
# Method-2 : using plt.subplots() : good for large numer of subplots
```

# Method-1 : using plt.subplot() : good for less number of plots

```python
# plt.subplot(nrows, ncols, index)
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
digits = load_digits()


mobile_company = ["Realme", "Samsung", "Nokia", "Lava", "Apple", "IQoo", "OnePlus"]
market_share = [15, 25, 2, 3, 30, 5, 20]

food_items = ["idli", "Dosa", "Vada", "Sambhar", "Puri"]
like_count = [2, 7, 3, 10, 12]


plt.figure(figsize = (8,6))

plt.subplot(2,3,1)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,2)
plt.barh(food_items, like_count, color = ["red", "blue", "orange", "green","purple"])


plt.subplot(2,3,3)
plt.imshow(digits.images[4])

plt.subplot(2,3,4)
plt.barh(food_items, like_count, color = ["red", "blue", "orange", "green","purple"])


plt.subplot(2,3,5)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,6)
plt.barh(food_items, like_count, color = ["red", "blue", "orange", "green","purple"])


plt.tight_layout() # use this if subplots are overlapping
plt.show()
```

# Method-2 : using plt.subplots() : good for large numer of subplots

```python
Dinner_dream_thali = ["Biryani", "Chicken", "Rusgulla", "ROTI", "Lassi"]
percentage = [30, 20, 10, 20, 20]

food_items = ["idli", "Dosa", "Vada", "Sambhar", "Puri"]
like_count = [2, 7, 3, 10, 12]


fig, axes = plt.subplots(2,3, figsize= (8,8)) # (nrows , nclos, full image size)
titles = ["Deepu", "Ruchitha", "akshara", "sandhya", "sreeja","pavan"]

print(axes)
axes_id = axes.ravel()
print(axes_id)

for title, ax in zip(titles, axes.ravel()):
    if title == "sandhya":
        ax.barh(food_items, like_count)
        ax.set_title(title)
    
    else:
        ax.pie(percentage, labels = Dinner_dream_thali, autopct = '%1.1f%%')
        ax.set_title(title)

# plt.tight_layout()
plt.show()
```

# 3D charts in matplotlib

```python
import numpy as np

z = np.linspace(0, 10, 100)
x = np.sin(z)
y = np.cos(z)

fig = plt.figure()
ax = fig.add_subplot(111, projection = '3d')
ax.scatter(x,y,z)
```
